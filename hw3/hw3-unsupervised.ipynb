Сделайте кластеризацию данных методом `KMeans` (собственная имплементация и готовая реализация), обучив модель на данных со сниженной за счет PCA размерностью. В данном случае мы подскажем, что нужно искать именно 6 кластеров, но в общем случае мы не будем знать, сколько кластеров надо искать.

Параметры:

- **n_clusters** = n_classes (число уникальных меток целевого класса)
- **n_init** = 100
- **random_state** = RANDOM_STATE (для воспроизводимости результата)

Остальные параметры со значениями по умолчанию.

````python
class MyKMeans:
    def __init__(self, n_clusters=6, n_init=10, max_iter=300, random_state=None):
        self.n_clusters = n_clusters
        self.n_init = n_init
        self.max_iter = max_iter
        self.random_state = random_state

    def fit(self, X):
        best_inertia = np.inf
        rng = np.random.default_rng(self.random_state)

        for _ in range(self.n_init):
            centroids = X[rng.choice(X.shape[0], self.n_clusters, replace=False)]

            for _ in range(self.max_iter):
                distances = np.linalg.norm(X[:, np.newaxis] - centroids, axis=2)
                labels = np.argmin(distances, axis=1)

                new_centroids = np.array([X[labels==i].mean(axis=0) if np.any(labels==i) else centroids[i]
                                          for i in range(self.n_clusters)])

                if np.allclose(new_centroids, centroids):
                    break
                centroids = new_centroids

            inertia = np.sum((X - centroids[labels])**2)
            if inertia < best_inertia:
                best_inertia = inertia
                self.labels_ = labels
                self.cluster_centers_ = centroids

my_kmeans = MyKMeans(n_clusters=n_classes, n_init=10, random_state=RANDOM_STATE)
my_kmeans.fit(X_pca)
my_labels = my_kmeans.labels_

kmeans = KMeans(n_clusters=n_classes, n_init=100, random_state=RANDOM_STATE)
kmeans.fit(X_pca)
sklearn_labels = kmeans.labels_

my_inertia = np.sum((X_pca - my_kmeans.cluster_centers_[my_labels])**2)
print("Inertia собственной реализации:", my_inertia)

# Inertia sklearn
sklearn_inertia = kmeans.inertia_
print("Inertia sklearn:", sklearn_inertia)
````
